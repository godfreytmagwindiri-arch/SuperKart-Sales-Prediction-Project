# **Problem Statement**

## Business Context

A sales forecast is a prediction of future sales revenue based on historical data, industry trends, and the status of the current sales pipeline. Businesses use the sales forecast to estimate weekly, monthly, quarterly, and annual sales totals. A company needs to make an accurate sales forecast as it adds value across an organization and helps the different verticals to chalk out their future course of action.

Forecasting helps an organization plan its sales operations by region and provides valuable insights to the supply chain team regarding the procurement of goods and materials. An accurate sales forecast process has many benefits which include improved decision-making about the future and reduction of sales pipeline and forecast risks. Moreover, it helps to reduce the time spent in planning territory coverage and establish benchmarks that can be used to assess trends in the future.

## Objective

SuperKart is a retail chain operating supermarkets and food marts across various tier cities, offering a wide range of products. To optimize its inventory management and make informed decisions around regional sales strategies, SuperKart wants to accurately forecast the sales revenue of its outlets for the upcoming quarter.

To operationalize these insights at scale, the company has partnered with a data science firm—not just to build a predictive model based on historical sales data, but to develop and deploy a robust forecasting solution that can be integrated into SuperKart’s decision-making systems and used across its network of stores.

## Data Description

The data contains the different attributes of the various products and stores.The detailed data dictionary is given below.

- **Product_Id** - unique identifier of each product, each identifier having two letters at the beginning followed by a number.
- **Product_Weight** - weight of each product
- **Product_Sugar_Content** - sugar content of each product like low sugar, regular and no sugar
- **Product_Allocated_Area** - ratio of the allocated display area of each product to the total display area of all the products in a store
- **Product_Type** - broad category for each product like meat, snack foods, hard drinks, dairy, canned, soft drinks, health and hygiene, baking goods, bread, breakfast, frozen foods, fruits and vegetables, household, seafood, starchy foods, others
- **Product_MRP** - maximum retail price of each product
- **Store_Id** - unique identifier of each store
- **Store_Establishment_Year** - year in which the store was established
- **Store_Size** - size of the store depending on sq. feet like high, medium and low
- **Store_Location_City_Type** - type of city in which the store is located like Tier 1, Tier 2 and Tier 3. Tier 1 consists of cities where the standard of living is comparatively higher than its Tier 2 and Tier 3 counterparts.
- **Store_Type** - type of store depending on the products that are being sold there like Departmental Store, Supermarket Type 1, Supermarket Type 2 and Food Mart
- **Product_Store_Sales_Total** - total revenue generated by the sale of that particular product in that particular store

# **Outlier Detection and Analysis Using the Interquartile Range (IQR) Method**

````python
# ============================================================
# OUTLIER DETECTION - IQR METHOD
# ============================================================

outlier_results = []

for column in numeric_cols:

    series = data[column].dropna()

    if series.empty:
        continue

    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = series[
        (series < lower_bound) |
        (series > upper_bound)]

    outlier_results.append({
        "Variable": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier %":
            len(outliers) / len(series) * 100})

outlier_summary = pd.DataFrame(outlier_results)

print("POTENTIAL OUTLIER SUMMARY")
display(outlier_summary.round(2))
````

# **Statistical Significance of Sales Differences Across Categorical Features (ANOVA and Kruskal–Wallis Tests)**

````python
# ============================================================
# ANOVA AND KRUSKAL-WALLIS TESTS
# ============================================================

statistical_results = []

for feature in cat_col:

    groups = [group[target_col]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .values
        for _, group in data.groupby(feature,observed=True,dropna=False)]

    groups = [g for g in groups if len(g) >= 2]

    if len(groups) < 2:
        continue

    if all(np.all(g == g[0])
        for g in groups):
        continue

    try:
        anova_stat, anova_p = f_oneway(*groups)
    except ValueError:
        anova_stat, anova_p = np.nan, np.nan

    try:
        kruskal_stat, kruskal_p = kruskal(*groups)
    except ValueError:
        kruskal_stat, kruskal_p = np.nan, np.nan

    statistical_results.append({"Feature": feature,
        "Number of Groups": len(groups),
        "ANOVA p-value": anova_p,
        "Kruskal p-value": kruskal_p})

statistical_report = pd.DataFrame(statistical_results)

display(statistical_report.round(5))
````

# **Observations**

2.	Kruskal–Wallis: Evaluates differences in sales distributions using ranks, providing an alternative when ANOVA assumptions are questionable.

# **Associations Between Categorical Features Using Chi-Square Tests and Cramér's V**

````python
from itertools import combinations
from scipy.stats import chi2_contingency
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Exclude extremely high-cardinality variables
eligible_cat = [col for col in cat_col
    if 2 <= data[col].nunique(dropna=False) <= 15]

categorical_associations = []

for col1, col2 in combinations(eligible_cat, 2):
    contingency_table = pd.crosstab(
        data[col1].fillna("Missing"),
        data[col2].fillna("Missing"))

    if min(contingency_table.shape) < 2:
        continue

    print(f"\nCROSSTAB: {col1} vs {col2}")
    display(contingency_table)

    plt.figure(figsize=(10, 6))
    sns.heatmap(
        contingency_table,
        annot=True,
        fmt="d",
        cmap="Blues")
    plt.title(f"{col1} vs {col2}")
    plt.tight_layout()
    plt.show()

    chi2, p_value, dof, expected = chi2_contingency(contingency_table)

    n = contingency_table.to_numpy().sum()
    r, k = contingency_table.shape

    cramers_v = np.sqrt((chi2 / n) / min(r - 1, k - 1))

    categorical_associations.append({
        "Variable 1": col1,
        "Variable 2": col2,
        "Chi-Square p-value": p_value,
        "Cramer's V": cramers_v,
        "Minimum Expected Count": expected.min()})

categorical_association_report = pd.DataFrame(categorical_associations)

display(categorical_association_report.round(4))
````

# **Observations**

1.	Categorical Relationships: Cross-tabulations and heatmaps reveal how product and store characteristics are distributed across different categories.

2.	Chi-Square Test: A p-value below 0.05 indicates a statistically significant association between two categorical variables.

3.	Association Strength: Cramér's V measures the strength of categorical associations, ranging from 0 (no association) to 1 (strongest possible association).

# **Summary of Numerical Feature Correlations with SuperKart Sales**

````python
# ============================================================
# NUMERICAL BIVARIATE SUMMARY
# ============================================================

numeric_summary = []

for feature in numerical_features:

    temp = data[[feature, target_col]].replace([np.inf, -np.inf], np.nan).dropna()

    if (len(temp) < 3
        or temp[feature].nunique() < 2
        or temp[target_col].nunique() < 2):
        continue

    numeric_summary.append({
        "Feature": feature,
        "Observations": len(temp),
        "Pearson Correlation":
            temp[feature].corr(temp[target_col]),
        "Spearman Correlation":
            temp[feature].corr(
                temp[target_col],
                method="spearman")})

numeric_bivariate_summary = pd.DataFrame(numeric_summary)

numeric_bivariate_summary["Absolute Pearson"] = (numeric_bivariate_summary[
        "Pearson Correlation"].abs())

numeric_bivariate_summary = (numeric_bivariate_summary
    .sort_values("Absolute Pearson",ascending=False))

display(numeric_bivariate_summary.round(4))
````

# **Variance Inflation Factor (VIF) Analysis for Multicollinearity Assessment**

````python
# ============================================================
# VARIANCE INFLATION FACTOR (VIF)
# ============================================================

vif_data = df_multi[numerical_features].copy()

# Replace infinite values and impute missing values
vif_data = vif_data.replace([np.inf, -np.inf], np.nan)

vif_data = vif_data.loc[:,vif_data.notna().any()]

vif_data = vif_data.fillna(vif_data.median())

# Remove constant variables
vif_data = vif_data.loc[:,vif_data.nunique() > 1]

if vif_data.shape[1] >= 2:

    # Standardise for numerical stability
    scaled_vif = StandardScaler().fit_transform(vif_data)

    vif_results = pd.DataFrame({"Feature": vif_data.columns,
        "VIF": [variance_inflation_factor(scaled_vif,i)
            for i in range(scaled_vif.shape[1])]})

    display(vif_results.sort_values("VIF", ascending=False).round(3))

else:
    print("Insufficient predictors for VIF analysis.")
````

````python
# ============================================================
# PRINCIPAL COMPONENT ANALYSIS
# ============================================================

pca_features = df_multi[numerical_features].copy()

pca_features = pca_features.replace([np.inf, -np.inf], np.nan)

pca_features = pca_features.loc[:,pca_features.notna().any()]

pca_features = pca_features.fillna(pca_features.median())

pca_features = pca_features.loc[:,pca_features.nunique() > 1]

if pca_features.shape[1] >= 2:

    scaled_features = StandardScaler().fit_transform(pca_features)

    pca = PCA(n_components=2)

    principal_components = pca.fit_transform(scaled_features)

    pca_df = pd.DataFrame(principal_components,columns=["PC1", "PC2"],index=pca_features.index)

    pca_df[target_col] = df_multi.loc[pca_df.index, target_col]

    plt.figure(figsize=(10, 7))

    sns.scatterplot(data=pca_df,x="PC1",y="PC2",hue=target_col,palette="viridis",alpha=0.7,legend=False)

    plt.title("SuperKart - PCA of Numerical Predictors")

    plt.tight_layout()
    plt.show()

    print("Explained Variance Ratio:")
    print(pca.explained_variance_ratio_)

    print("Total Explained Variance:",round(pca.explained_variance_ratio_.sum() * 100,2),"%")

else:
    print("Insufficient numerical features for PCA.")
````

# **Preprocessing pipeline – fit on training data only**

````python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error, mean_absolute_percentage_error
import numpy as np
import pandas as pd

cat_cols = X_train.select_dtypes(include=['object','category','string','bool']).columns.tolist()
num_cols = X_train.select_dtypes(include=np.number).columns.tolist()

def make_preprocessor():
    return ColumnTransformer([
      ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')),
                        ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))]), cat_cols),
      ('num', Pipeline([('imputer', SimpleImputer(strategy='median'))]), num_cols)
    ], remainder='drop')

print('Categorical columns:',cat_cols)
print('Numeric columns:',num_cols)
print('No standalone encoding: each model pipeline learns training-only preprocessing.')
````

3.	Numerical missing values: Median imputation addresses missing numerical values and is less sensitive to extreme observations than mean imputation.

4.	Categorical encoding: OneHotEncoder converts categorical variables into numerical representations, enabling their use in Decision Tree and Gradient Boosting regression models.

5.	Handling unseen categories: handle_unknown='ignore' allows the model to process categories encountered during testing that were not present during training, improving robustness.

6.	Consistent preprocessing: ColumnTransformer applies different preprocessing techniques to categorical and numerical features within a unified framework.

7.	Prevention of data leakage: Integrating preprocessing into model pipelines ensures that imputation and encoding parameters are learned exclusively from training data during model fitting and cross-validation.

8.	Feature retention: remainder='drop' excludes variables not explicitly assigned to a preprocessing transformation, maintaining control over the features entering the models.

9.	Model compatibility: The preprocessing framework supports both Decision Tree and Gradient Boosting models, enabling consistent and fair performance comparisons.

# **9. Evaluation – training and testing performance**

````python
# ============================================================
# STAGE 10: BASELINE MODEL PERFORMANCE EVALUATION
# DECISION TREE VS GRADIENT BOOSTING
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (r2_score,
    mean_squared_error,
    mean_absolute_error,
    mean_absolute_percentage_error)

# ------------------------------------------------------------
# 1. DEFINE MODEL EVALUATION FUNCTION
# ------------------------------------------------------------

def model_performance_regression(model, predictors, target):

    predictions = model.predict(predictors)

    rmse = np.sqrt(mean_squared_error(target, predictions))
    mae = mean_absolute_error(target, predictions)
    r2 = r2_score(target, predictions)

    # MAPE is undefined if actual sales contain zeros
    actual = np.asarray(target)

    mape = (mean_absolute_percentage_error(target, predictions) * 100
        if np.all(actual != 0)
        else np.nan)

    return {"RMSE": rmse,
        "MAE": mae,
        "R-squared": r2,
        "MAPE (%)": mape}
# ------------------------------------------------------------
# 2. EVALUATE TRAINING AND TESTING PERFORMANCE
# ------------------------------------------------------------

models = {"Decision Tree": dtree,
    "Gradient Boosting": gb_pipeline}

results = []

for name, model in models.items():

    for split, X_part, y_part in [
        ("Train", X_train, y_train),
        ("Test", X_test, y_test)]:

        metrics = model_performance_regression(
            model, X_part, y_part)

        results.append({
            "Model": name,
            "Dataset": split,
            **metrics})
# ------------------------------------------------------------
# 3. CONSOLIDATED PERFORMANCE COMPARISON
# ------------------------------------------------------------
initial_comparison = pd.DataFrame(results)

print("\nBASELINE MODEL PERFORMANCE COMPARISON")
print("=" * 55)

display(initial_comparison.round(4))
# ------------------------------------------------------------
# 4. COMPARE TRAINING AND TESTING R-SQUARED
# ------------------------------------------------------------

r2_comparison = initial_comparison.pivot(
    index="Model",
    columns="Dataset",
    values="R-squared")

r2_comparison["R2_Gap"] = (r2_comparison["Train"] - r2_comparison["Test"])

print("\nTRAINING VS TESTING R-SQUARED")
display(r2_comparison.round(4))
# ------------------------------------------------------------
# 5. IDENTIFY BEST BASELINE MODEL (DESCRIPTIVE ONLY)
# ------------------------------------------------------------

test_results = initial_comparison[
    initial_comparison["Dataset"] == "Test"
].copy()

test_results = test_results.sort_values("RMSE")

print("\nTEST PERFORMANCE RANKING")
display(test_results.round(4))

best_baseline = test_results.iloc[0]

print("\nLowest Test RMSE:")
print(best_baseline["Model"])
print("RMSE:", round(best_baseline["RMSE"], 4))
print("MAE:", round(best_baseline["MAE"], 4))
print("R²:", round(best_baseline["R-squared"], 4))

print("\nNote: Final model selection should use "
    "training-only cross-validation.")
# ------------------------------------------------------------
# 6. VISUALIZE TRAINING VS TESTING PERFORMANCE
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

metrics = ["RMSE", "MAE", "R-squared"]

for ax, metric in zip(axes, metrics):

    comparison = initial_comparison.pivot(index="Model",
        columns="Dataset",values=metric)

    comparison.plot(kind="bar",ax=ax,rot=0)

    ax.set_title(f"Baseline Model Comparison - {metric}")
    ax.set_xlabel("Model")
    ax.set_ylabel(metric)
    ax.legend(title="Dataset")

plt.tight_layout()
plt.show()
````

````python
# ============================================================
# STAGE 15: RESIDUAL ANALYSIS – SUPERKART
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Calculate residuals
actual = np.asarray(y_test)
predicted = final_model.predict(X_test)
residuals = actual - predicted

# 2. Residual summary
residual_summary = pd.DataFrame({'Mean Error': [np.mean(residuals)],
    'Median Error': [np.median(residuals)],
    'Std Error': [np.std(residuals)],
    'MAE': [np.mean(np.abs(residuals))],
    'RMSE': [np.sqrt(np.mean(residuals**2))],
    'Max Absolute Error': [np.max(np.abs(residuals))]})

print("RESIDUAL ANALYSIS SUMMARY")
display(residual_summary.round(4))

# 3. Visualize prediction errors
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Residuals vs predicted sales
axes[0].scatter(predicted, residuals, alpha=0.4)
axes[0].axhline(0, linestyle='--')
axes[0].set(title='Residuals vs Predicted Sales',
            xlabel='Predicted Sales', ylabel='Residuals')

# Residual distribution
axes[1].hist(residuals, bins=30, edgecolor='black')
axes[1].axvline(0, linestyle='--')
axes[1].set(title='Residual Distribution',
            xlabel='Residual Error', ylabel='Frequency')

# Actual vs predicted sales
axes[2].scatter(actual, predicted, alpha=0.4)
lims = [min(actual.min(), predicted.min()),
        max(actual.max(), predicted.max())]
axes[2].plot(lims, lims, '--')
axes[2].set(title='Actual vs Predicted Sales',
            xlabel='Actual Sales', ylabel='Predicted Sales')

plt.suptitle(f'SuperKart – {final_model_name}: Residual Analysis')
plt.tight_layout()
plt.show()
````

# **17 – Model Generalization and Overfitting Assessment**

````python
# ============================================================
# STAGE 17: MODEL GENERALIZATION & OVERFITTING
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Evaluate training and testing performance
results = []
for split, X, y in [('Train', X_train, y_train), ('Test', X_test, y_test)]:
    pred = final_model.predict(X)
    results.append({
        'Dataset': split,
        'R²': r2_score(y, pred),
        'RMSE': np.sqrt(mean_squared_error(y, pred)),
        'MAE': mean_absolute_error(y, pred)})

generalization = pd.DataFrame(results)
train, test = generalization.iloc[0], generalization.iloc[1]

# Extract selected model's cross-validation RMSE
search = dt_search if final_model_name in ['Decision Tree', 'Tuned Decision Tree'] else gb_search
cv_rmse = -search.best_score_

# Display performance and generalization gaps
display(generalization.round(4))
print(f"CV RMSE: {cv_rmse:.4f}")
print(f"R² Gap (Train - Test): {train['R²'] - test['R²']:.4f}")
print(f"RMSE Gap (Test - Train): {test['RMSE'] - train['RMSE']:.4f}")
print(f"Test - CV RMSE Gap: {test['RMSE'] - cv_rmse:.4f}")

# Visualize training vs testing
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, metric in zip(axes, ['R²', 'RMSE', 'MAE']):
    ax.bar(generalization['Dataset'], generalization[metric])
    ax.set(title=f'{metric} – Train vs Test', ylabel=metric)

plt.suptitle(f'SuperKart – {final_model_name}: Generalization')
plt.tight_layout()
plt.show()
````

# **Observations**

1.	Training versus testing R²: A large positive gap suggests potential overfitting, while similar values provide evidence of more consistent predictive performance.

2.	RMSE generalization gap: A substantially higher testing RMSE than training RMSE may indicate that the model fits training observations more accurately than unseen data.

3.	Cross-validation RMSE: Provides a benchmark for assessing whether the final test error is broadly consistent with performance observed during model selection.

4.	Model stability: Similar cross-validation and testing RMSE values support confidence in generalization, although they do not guarantee future performance.

5.	Underfitting: Poor performance on both training and testing datasets may indicate that the model is insufficiently flexible or that available features have limited predictive information.

6.	Business implications: A model with stable performance is more suitable for further operational validation in SuperKart sales forecasting and inventory planning.

7.	Important: There is no universal R² or RMSE gap threshold that proves overfitting. Interpret these results in relation to sales variability, baseline performance and cross-validation uncertainty. Since Stage 14 already evaluated the held-out test data, use this stage as a diagnostic assessment rather than another opportunity to tune the model.

````python
%%writefile /content/backend_files/app.py
# ============================================================
# SUPERKART – FLASK BACKEND API
# GOOGLE COLAB / DOCKER DEPLOYMENT
# ============================================================

from flask import Flask, request, jsonify
from pathlib import Path
import pandas as pd
import numpy as np
import joblib
import json

app = Flask(__name__)
app.config["MAX_CONTENT_LENGTH"] = 5 * 1024 * 1024

BASE_DIR = Path(__file__).resolve().parent

# Load fitted model and feature schema (pointing to superkart_model.joblib)
model = joblib.load(BASE_DIR / "superkart_model.joblib")

with open(BASE_DIR / "feature_schema.json", encoding="utf-8") as f:
    schema = json.load(f)

FEATURES = schema["features"]
NUMERIC = schema["numeric_features"]
CATEGORICAL = schema["categorical_features"]

# Validate prediction inputs
def prepare_input(df):
    missing = [c for c in FEATURES if c not in df.columns]
    if missing:
        raise ValueError(f"Missing predictors: {missing}")

    df = df[FEATURES].copy()

    for col in NUMERIC:
        converted = pd.to_numeric(df[col], errors="coerce")
        invalid = df[col].notna() & converted.isna()
        if invalid.any():
            raise ValueError(f"Invalid numeric values: {col}")
        df[col] = converted

    for col in CATEGORICAL:
        df[col] = df[col].astype("string").astype(object)
        df[col] = df[col].where(df[col].notna(), np.nan)

    return df

# Home endpoint
@app.get("/")
def home():
    return jsonify({
        "application": "SuperKart Sales Prediction API",
        "status": "running"})

# Health check
@app.get("/health")
def health():
    return jsonify({"status": "healthy"})

# Feature schema
@app.get("/v1/schema")
def show_schema():
    return jsonify(schema)

# Single sales prediction
@app.post("/v1/sales")
def predict_sales():
    try:
        payload = request.get_json(silent=True)
        if not isinstance(payload, dict):
            return jsonify({"error": "Expected JSON object"}), 400

        df = prepare_input(pd.DataFrame([payload]))
        prediction = float(model.predict(df)[0])

        if not np.isfinite(prediction):
            raise ValueError("Non-finite prediction")

        return jsonify({
            "Predicted_Sales": round(prediction, 2)})

    except ValueError as e:
        return jsonify({"error": str(e)}), 400
    except Exception:
        app.logger.exception("Prediction failed")
        return jsonify({"error": "Prediction failed"}), 500

# Batch CSV predictions
@app.post("/v1/salesbatch")
def predict_batch():
    try:
        file = request.files.get("file")
        if file is None:
            return jsonify({"error": "CSV file required"}), 400

        if not file.filename.lower().endswith(".csv"):
            return jsonify({"error": "Upload a CSV file"}), 400

        df = pd.read_csv(file)
        if df.empty or len(df) > 1000:
            raise ValueError("CSV must have 1–1000 rows")

        prepared = prepare_input(df)
        predictions = model.predict(prepared)

        if not np.isfinite(predictions).all():
            raise ValueError("Non-finite predictions")

        df["Predicted_Sales"] = np.round(predictions, 2)

        return jsonify({
            "total_records": len(df),
            "predictions": json.loads(df.to_json(orient="records"))})

    except (ValueError, pd.errors.ParserError) as e:
        return jsonify({"error": str(e)}), 400
    except Exception:
        app.logger.exception("Batch prediction failed")
        return jsonify({"error": "Batch prediction failed"}), 500

if __name__ == "__main__":
    app.run(host="0.0.0.0", port=7860, debug=False)
````

# **D. Flask REST API: full production-oriented example**

````python
"""SuperKart inference API. Load only trusted local joblib artifacts."""
import json
import logging
import os
from pathlib import Path
import joblib
import numpy as np
import pandas as pd
from flask import Flask, jsonify, request

# Fallback to current working directory when running inside interactive notebooks where __file__ is undefined
try:
    BASE = Path(__file__).resolve().parent
except NameError:
    BASE = Path.cwd()

MODEL_PATH = Path(os.getenv("MODEL_PATH", str(BASE / "models/superkart_final_model.joblib")))
SCHEMA_PATH = Path(os.getenv("SCHEMA_PATH", str(BASE / "models/feature_schema.json")))
MAX_ROWS = int(os.getenv("MAX_BATCH_ROWS", "1000"))
app = Flask(__name__)
app.config["MAX_CONTENT_LENGTH"] = 5 * 1024 * 1024  # 5 MB
logging.basicConfig(level=logging.INFO)
model = joblib.load(MODEL_PATH)  # Only load trusted artifacts.
schema = json.loads(SCHEMA_PATH.read_text(encoding="utf-8"))
FEATURES = schema["features"]
NUMERIC = set(schema["numeric_features"])
CATEGORICAL = set(schema["categorical_features"])
PERISHABLES = {"Dairy", "Meat", "Fruits and Vegetables", "Breakfast", "Breads", "Seafood"}


def prepare(records, raw=False):
    if not isinstance(records, list) or not records or len(records) > MAX_ROWS:
        raise ValueError(f"Provide 1 to {MAX_ROWS} records")
    if not all(isinstance(r, dict) for r in records):
        raise ValueError("Every record must be a JSON object")
    df = pd.DataFrame(records)
    if raw:
        required_raw = ["Product_Id", "Store_Establishment_Year", "Product_Type"]
        absent = [k for k in required_raw if k not in df.columns]
        if absent:
            raise ValueError(f"Missing raw features: {absent}")
        df["Product_Id_char"] = df["Product_Id"].astype("string").str[:2]
        year = pd.to_numeric(df["Store_Establishment_Year"], errors="coerce")
        df["Store_Age_Years"] = schema["reference_year"] - year
        df["Product_Type_Category"] = np.where(df["Product_Type"].isin(PERISHABLES), "Perishables", "Non Perishables")
    if "Product_Sugar_Content" in df:
        df["Product_Sugar_Content"] = df["Product_Sugar_Content"].replace({"reg": "Regular", "Reg": "Regular"})
    missing = [c for c in FEATURES if c not in df.columns]
    if missing:
        raise ValueError(f"Missing model features: {missing}")
    df = df.loc[:, FEATURES].copy()
    for c in NUMERIC:
        df[c] = pd.to_numeric(df[c], errors="coerce")
        if df[c].notna().sum() == 0:
            raise ValueError(f"Feature {c} has no valid numeric values")
    for c in CATEGORICAL:
        df[c] = df[c].astype("object").where(df[c].notna(), np.nan)
    return df


def infer(records, raw=False):
    X = prepare(records, raw)
    preds = model.predict(X)
    if not np.isfinite(preds).all():
        raise ValueError("Non-finite prediction returned")
    return [round(float(p), 4) for p in preds]

@app.get("/health")
def health():
    return jsonify({"status": "ok", "model_loaded": True})

@app.get("/v1/schema")
def show_schema():
    return jsonify(schema)

@app.post("/v1/predict")
def predict():
    try:
        payload = request.get_json(silent=True)
        if not isinstance(payload, dict):
            raise ValueError("Expected JSON object containing model features")
        result = infer([payload])[0]
        return jsonify({"prediction": result, "model": schema.get("model_name", "SuperKart")})
    except ValueError as exc:
        return jsonify({"error": str(exc)}), 400
    except Exception:
        app.logger.exception("Inference error")
        return jsonify({"error": "Prediction service error"}), 500

@app.post("/v1/predict/raw")
def predict_raw():
    try:
        payload = request.get_json(silent=True)
        if not isinstance(payload, dict):
            raise ValueError("Expected JSON object with raw product/store features")
        return jsonify({"prediction": infer([payload], raw=True)[0]})
    except ValueError as exc:
        return jsonify({"error": str(exc)}), 400
    except Exception:
        app.logger.exception("Raw inference error")
        return jsonify({"error": "Prediction service error"}), 500

@app.post("/v1/predictbatch")
def predict_batch():
    try:
        if "file" not in request.files:
            raise ValueError("Upload a CSV using multipart field named 'file'")
        file = request.files["file"]
        if not file.filename.lower().endswith(".csv"):
            raise ValueError("Only CSV files are supported")
        df = pd.read_csv(file)
        preds = infer(df.to_dict(orient="records"), raw=request.args.get("raw") == "true")
        return jsonify({"count": len(preds), "predictions": preds})
    except (ValueError, pd.errors.ParserError, UnicodeDecodeError) as exc:
        return jsonify({"error": str(exc)}), 400
    except Exception:
        app.logger.exception("Batch inference error")
        return jsonify({"error": "Batch prediction service error"}), 500

````

# **F. Streamlit frontend: responsive single-prediction form and CSV batch upload**

````python
"""Streamlit frontend for the SuperKart Flask prediction API."""
import os
from io import BytesIO
import pandas as pd
import requests
import streamlit as st

st.set_page_config(page_title="SuperKart Sales Predictor", layout="wide")
st.title("SuperKart | Sales Prediction")
API = os.getenv("BACKEND_URL", "http://localhost:7860").rstrip("/")
TIMEOUT = 30

@st.cache_data(ttl=300)
def get_schema(api):
    r = requests.get(f"{api}/v1/schema", timeout=TIMEOUT)
    r.raise_for_status()
    return r.json()

try:
    schema = get_schema(API)
except requests.RequestException as e:
    st.error(f"Cannot reach backend: {e}")
    st.stop()

st.caption("Predictions are model estimates, not guaranteed future sales. Units match training target.")
tab1, tab2 = st.tabs(["Single prediction", "Batch CSV predictions"])
with tab1:
    st.subheader("Enter product and store characteristics")
    with st.form("predict_form"):
        values = {}
        left, right = st.columns(2)
        for i, name in enumerate(schema["features"]):
            with (left if i % 2 == 0 else right):
                if name in schema["numeric_features"]:
                    default = float(schema.get("numeric_defaults", {}).get(name, 0))
                    values[name] = st.number_input(name, value=default, format="%.4f")
                else:
                    options = schema.get("categorical_values", {}).get(name, [])
                    options = [str(x) for x in options if x is not None]
                    if options:
                        values[name] = st.selectbox(name, options)
                    else:
                        values[name] = st.text_input(name)
        submitted = st.form_submit_button("Predict sales", type="primary")
    if submitted:
        try:
            r = requests.post(f"{API}/v1/predict", json=values, timeout=TIMEOUT)
            r.raise_for_status()
            st.metric("Estimated sales", f"{r.json()['prediction']:,.2f}")
        except (requests.RequestException, ValueError, KeyError) as e:
            st.error(f"Prediction failed: {e}")
            if 'r' in locals() and not r.ok:
                st.code(r.text)

with tab2:
    st.subheader("Upload a CSV with the required features")
    st.write("Required columns:", ", ".join(schema["features"]))
    uploaded = st.file_uploader("CSV file", type=["csv"])
    if uploaded is not None and st.button("Run batch inference"):
        try:
            content = uploaded.getvalue()
            source = pd.read_csv(BytesIO(content))
            r = requests.post(f"{API}/v1/predictbatch", files={"file": (uploaded.name, content, "text/csv")}, timeout=120)
            r.raise_for_status()
            predictions = r.json()["predictions"]
            if len(predictions) != len(source):
                raise ValueError("Prediction count does not match CSV rows")
            source["Predicted_Sales_Total"] = predictions
            st.dataframe(source, use_container_width=True)
            st.download_button("Download predictions", source.to_csv(index=False), file_name="superkart_predictions.csv", mime="text/csv")
        except (requests.RequestException, ValueError, KeyError) as e:
            st.error(f"Batch prediction failed: {e}")
            if 'r' in locals() and not r.ok:
                st.code(r.text)
````
